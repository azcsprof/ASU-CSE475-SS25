```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
```

```python
# Load the penguins dataset
penguins = sns.load_dataset("penguins")
```

```python
# --- GENERAL STRUCTURE + SUMMARY STATS ---

print("📌 First 5 rows:")
print(penguins.head())

print("\n📌 Shape of the dataset:")
print("✅ shape_data:", penguins.shape)

print("\n📌 Data types and non-null counts:")
print(penguins.info())

print("\n📌 Summary statistics for numeric columns:")
print(penguins.describe())

print("\n📌 Null values per column:")
print(penguins.isnull().sum())

print("\n📌 Number of duplicate rows:")
print(penguins.duplicated().sum())
```

```python
# --- FEATURE-LEVEL ANALYSIS ---

print("\n📌 Number of unique values per column:")
for col in penguins.columns:
    unique_count = penguins[col].nunique(dropna=True)
    print(f"{col}: {unique_count}")

print("\n📌 Value counts for categorical features:")
for col in penguins.select_dtypes(include='object').columns:
    print(f"\n▶ {col}:")
    print(penguins[col].value_counts(dropna=False))
```

```python
import matplotlib.pyplot as plt

numeric_cols = penguins.select_dtypes(include='number').columns
penguins[numeric_cols].hist(bins=20, figsize=(12, 8))
plt.suptitle("📊 Distribution of Numeric Features")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
sns.heatmap(penguins.isnull(), cbar=False, cmap='viridis')
plt.title("📊 Missing Values Heatmap")
plt.show()
```

```python
# STEP: Calculate skewness and kurtosis for selected features

# Calculate the correlation matrix
correlations = penguins.corr(numeric_only=True)['body_mass_g'].sort_values(ascending=False)

# Let's try this for the top 5 predictive features (from earlier correlation step)
top_features = correlations[1:6].index.tolist()

# Replace 'df' with 'penguins' as per the notebook context
for feature in top_features:
    skew = penguins[feature].skew()
    kurt = penguins[feature].kurtosis()

    print(f"\n📌 Feature: {feature}")
    print(f"   Skewness: {skew:.2f} →", end=" ")
    if abs(skew) < 0.5:
        print("Fairly symmetric")
    elif skew > 0:
        print("Right-skewed")
    else:
        print("Left-skewed")

    print(f"   Kurtosis: {kurt:.2f} →", end=" ")
    if kurt < 3:
        print("Light-tailed (fewer outliers)")
    elif kurt > 3:
        print("Heavy-tailed (more outliers)")
    else:
        print("Approximately normal")
```

```python
# --- CORE DATA ANALYSIS TASKS ---
# 1. Max values
bill_length_max = penguins['bill_length_mm'].max()
flipper_length_max = penguins['flipper_length_mm'].max()
print("✅ bill_length_max:", bill_length_max)
print("✅ flipper_length_max:", flipper_length_max)
```

```python
# 2. Null counts
body_mass_null = penguins['body_mass_g'].isnull().sum()
flipper_length_null = penguins['flipper_length_mm'].isnull().sum()
species_null = penguins['species'].isnull().sum()
print("✅ body_mass_null:", body_mass_null)
print("✅ flipper_length_null:", flipper_length_null)
print("✅ species_null:", species_null)
```

```python
# 3. Mean body mass after dropping nulls
penguins_drop = penguins.dropna()
mean_body_mass_drop = penguins_drop['body_mass_g'].mean()
print("✅ mean_body_mass_drop:", mean_body_mass_drop)
```

```python
# 4. Mean body mass after filling nulls
penguins_fill = penguins.copy()
penguins_fill['body_mass_g'] = penguins_fill['body_mass_g'].fillna(penguins_fill['body_mass_g'].median())
mean_body_mass_fill = penguins_fill['body_mass_g'].mean()
print("✅ mean_body_mass_fill:", mean_body_mass_fill)
```

```python
# 5. Normalize body mass
scaler = MinMaxScaler()
penguins_fill['body_mass_norm'] = scaler.fit_transform(penguins_fill[['body_mass_g']])
mean_body_mass_norm = penguins_fill['body_mass_norm'].mean()
print("✅ mean_body_mass_norm:", mean_body_mass_norm)
```

```python
# 6. Classify penguins by size
def classify_mass(mass):
    if mass < 3500:
        return 'Small'
    elif mass < 5000:
        return 'Medium'
    else:
        return 'Large'

penguins_fill['mass_class'] = penguins_fill['body_mass_g'].apply(classify_mass)
small_count = (penguins_fill['mass_class'] == 'Small').sum()
medium_count = (penguins_fill['mass_class'] == 'Medium').sum()
large_count = (penguins_fill['mass_class'] == 'Large').sum()
print("✅ small_count:", small_count)
print("✅ medium_count:", medium_count)
print("✅ large_count:", large_count)
```

```python
# 7. Find index of largest penguin
max_index = penguins_fill['body_mass_g'].idxmax()
print("✅ max_index:", max_index)
```

```python
# 8. Extract metadata from that penguin
max_species = penguins_fill.loc[max_index, 'species']
max_island = penguins_fill.loc[max_index, 'island']
max_sex = penguins_fill.loc[max_index, 'sex']
print("✅ max_species:", max_species)
print("✅ max_island:", max_island)
print("✅ max_sex:", max_sex)
```

```python
# 9. Filter: Large penguins with flipper > 220
filtered = penguins_fill[
    (penguins_fill['mass_class'] == 'Large') &
    (penguins_fill['flipper_length_mm'] > 220)
]
print("✅ Filtered large penguins with flipper > 220:", filtered.shape[0])
```

```python
# 10. Group by island and report mean body mass
grouped_mass = penguins_fill.groupby('island')['body_mass_g'].mean()
print("✅ mean_body_mass_biscoe:", round(grouped_mass['Biscoe'], 2))
```

```python
# 11. Create a binary classification column: Label penguins as 'is_heavy' based on body mass threshold
penguins_fill['is_heavy'] = penguins_fill['body_mass_g'] > 5000
heavy_count = penguins_fill['is_heavy'].sum()
print("✅ heavy_count:", heavy_count)
```

```python
# 12. Filter for “deep-billed males” and count
deep_males = penguins_fill[
    (penguins_fill['bill_depth_mm'] > 18) &
    (penguins_fill['sex'] == 'Male')
]
print("✅ Count of deep-billed males:", deep_males.shape[0])
```

```python
# 13. Compare mean flipper length before and after filtering
# Show how filtering affects numeric summaries
mean_flipper_before = penguins_fill['flipper_length_mm'].mean()
filtered_df = penguins_fill[penguins_fill['bill_length_mm'] > 45]
mean_flipper_after = filtered_df['flipper_length_mm'].mean()
print("✅ mean_flipper_before:", round(mean_flipper_before, 2))
print("✅ mean_flipper_after (bill_length > 45):", round(mean_flipper_after, 2))
```

```python
# 14 – Calculate mode of a categorical column
# Categorical aggregation (not just means)
most_common_island = penguins_fill['island'].mode()[0]
print("✅ most_common_island:", most_common_island)
```

```python
# 15. Group by species and report average bill depth
grouped_bill_depth = penguins_fill.groupby('species')['bill_depth_mm'].mean()
print("✅ mean_bill_depth_adelie:", round(grouped_bill_depth['Adelie'], 2))
```
